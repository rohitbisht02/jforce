```python
import pandas as pd

```

```python
customer_df = pd.read_csv('customer_data.csv')
customer_df.head()
```

```python
customer_df.info()

```

```python
customer_df = customer_df.drop_duplicates().reset_index(drop=True)
customer_df.head()
```

```python
sales_df = pd.read_csv('sales.csv')
sales_df.head()
```

```python
sales_df.columns = sales_df.columns.str.strip().str.lower()
sales_df.head()
```

```python
sales_df['quantity_sold'] = pd.to_numeric(sales_df['quantity_sold'], errors='coerce')
sales_df = sales_df.dropna(subset=['quantity_sold']).reset_index(drop=True)
sales_df['quantity_sold'] = sales_df['quantity_sold'].astype(int)
sales_df.dtypes
```

```python
product_df = pd.read_csv('product.csv')
product_df.columns = product_df.columns.str.strip().str.lower()
product_df.head()
```

```python
product_df = product_df.drop_duplicates(subset=['product_id']).reset_index(drop=True)
product_df = product_df.dropna(subset=['product_id', 'price']).reset_index(drop=True)
product_df['price'] = pd.to_numeric(product_df['price'], errors='coerce')
product_df.head()
```

```python
sales_df = sales_df.merge(
    product_df[['product_id', 'price']].rename(columns={'price': 'unit_price'}),
    on='product_id',
    how='left'
)
sales_df['unit_price'] = pd.to_numeric(sales_df['unit_price'], errors='coerce')
sales_df.head()
```

```python
sales_df['total_sale_amount'] = sales_df['quantity_sold'] * sales_df['unit_price']
sales_df[['sale_id', 'product_id', 'quantity_sold', 'unit_price', 'total_sale_amount']].head()
```

```python
print('Sales shape:', sales_df.shape)
print('Missing unit_price count:', sales_df['unit_price'].isna().sum())
sales_df[['sale_id', 'product_id', 'quantity_sold', 'unit_price', 'total_sale_amount']].describe()
```

```python
customer_df.to_csv('customer_data_clean.csv', index=False)
sales_df.to_csv('sales_clean.csv', index=False)
product_df.to_csv('product_clean.csv', index=False)
```
